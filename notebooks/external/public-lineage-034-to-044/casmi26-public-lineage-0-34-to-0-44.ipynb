# How the public leaderboard went from 0.34 to 0.44 — a lineage map of the public notebooks

This is a map of which public notebook in **enveda-CASMI26-molecule-id-mass-spectra** was built on which, what each step changed, and what it scored.
It credits every author named in the sources.
It is not a ranking of who deserves the most credit; it is a record of lineage.


## How to read this

Every score below is labelled.
**Claimed by the author** means the number comes from the notebook's own title, markdown, or code comment.
**Measured when re-run verbatim** means the exact public notebook was re-submitted on 2026-10-09 and its public score read from the leaderboard; only three numbers in this whole map carry that label.
A few numbers are **board readings** — a team's public leaderboard position observed on a given date, not tied by the source to one specific notebook version.
Those are marked explicitly and are not notebook scores.
Where a step's own score is not stated anywhere, this map says "not stated." No score here is invented.

Three families run in parallel for most of this story, then merge:

- **The analog-propagation / two-ranker family**, descending from prvsiyan's baseline notebook.
- **Ahmed Berat Özer's own engine**, built separately (own FPNet, own ranker, own generation).
- **seyitkaangunes's fusion step**, which reciprocal-rank-sum blends the two families together, and on which almost everything above 0.38 is built.


## The root: analog propagation

**prvsiyan/analog-propagation-casmi-2026-baseline** is the lineage's root notebook: library matching plus mass-shifted analog propagation.
Its V33 revision reports a pop-prior-plus-forward-isomer-check-plus-gated-PubChem-tier variant scoring 0.340 twice, claimed by the author.
Its own table elsewhere shows that flooding the candidate pool with PubChem top-50 isomers by f.z, ungated, drops the score to 0.205 against 0.335 for the gated version — a documented null for naive pool expansion, claimed by the author.

From this root, two early forks matter:

- **flexonafft/casmi26-strong-retrieval-clean-ranking**, a fork of prvsiyan V17, with a byte-identical copy at **uninhibitedscholar/casmi26-strong-retrieval-replica-v23**.
- **evgendvorkin/enveda-casmi-2026** and **beraterolelk/0-336-sota-enveda-casmi26-analog-ranker**, the origin of the "two-ranker" blend (also associated with the handle megayak), which the notes say emits the lineage's rank-1 pick on 400 of 400 molecules.

## The merge: seyitkaangunes's fusion

This is the hinge of the whole story — the point where the two-ranker family and Ahmed Berat Özer's engine stop being separate lineages.

- **seyitkaangunes/casmi26-v4g-own-engine-rrf-fusion-lb-0-382**: claims 0.382. Fuses v4g with the prvsiyan/megayak two-ranker engine, adds ChEBI/LIPID MAPS to the pool ("BIO", claimed 0.341 → 0.350 alone) and a formula-mass analog index ("AFIX"). Reciprocal-rank-sum on the tautomer-canonical InChIKey-14. The author's own public alpha sweep at K=3: alpha 0.2 → 0.370, 0.35 → 0.378, 0.6 → 0.382, 1.0 → 0.366 (v4g alone claimed 0.366).
- **seyitkaangunes/casmi26-v4n-engine-fusion-union-lb-0-399**: claims 0.399. Same fusion on the v4n base (claimed 0.384), plus ICEBERG and GLACIER scoring of the fused top-40 ("union"), then a same-formula re-rank. The author's own steps on the v4g base: 0.366 → 0.382 (fusion) → 0.386 (union).
- **seyitkaangunes/casmi26-e5c-union-gl-a06**: the same engine plus union plus GLACIER at alpha 0.6, an intermediate step toward the 0.399 notebook.
- **ahmedberatozer/casmi26-v4q-inference** = v4n + e5c: the author absorbing the fusion idea back into the v4-series.
- **ahmedberatozer/casmi26-v4n-inference** = v4i with equal ICE/GLACIER weights and the fpnet_full1 bank.

## What the steps have in common

Reading the whole lineage, the gains cluster around six mechanisms, in roughly the order they appear:

1. **Engine fusion** — reciprocal-rank-sum blending of two independently built engines (the two-ranker/analog-propagation family and Ahmed Berat Özer's own engine), introduced by seyitkaangunes and then grafted onto almost every base that follows.
2. **Forward-model rescoring** — ICEBERG and GLACIER re-score the fused candidate list ("union"). This step was confounded for days by a Python 3.12-vs-3.13 RDKit wheel mismatch that silently zeroed out ICEBERG/GLACIER scoring in notebooks that didn't pin the right image or patch the wheel; this was documented independently by lszlst, skomuro, and ahmedberatozer's v6a.
3. **Popularity priors** — log1p-scaled PubChem SID/PMID/patent counts, added as a weighted term, first run on private arrays (wangpenghua) and then made usable by everyone once dmitriigluzdov released the counts as a public dataset.
4. **PubChem-join / PubChem-list ranking** — moving from fixed-slot interleaving of PubChem-only candidates to ranking them through the real engine's own features. This is the single largest documented step in the whole table (huseyinemreaksoy, +0.011) and the explicit mechanism behind the CLAW promotion rule (lszlst).
5. **Tail backfill and list-level dedup** — filling short candidate lists from DreaMS mass-window matches or ranked overflow, and deduplicating by the scorer's own tautomer-canonical identity rather than the raw library key (obstacledeveloper's V1.1, nursrijan's consensus_backfill, spark328/capwitf's second DreaMS pass).
6. **Rank-1 / library-hit shields** — locking or down-weighting forward-model scores when library confidence is already high. Evidence on this one is mixed: YahiaZakaria20 reports an offline validation gain from 0.63 to 0.89 from locking rank 1, but a board loss from 0.413 to 0.409 on the same change, because the locked structures are well-documented training compounds, not representative of the harder test classes.


## Licence note

Train.parquet is labelled CC BY-NC 4.0 but the hosts treat it as an open-source compilation of other open libraries; the file's own licence label is not read as a commercial-use bar. Test data carries its own, separate noncommercial-use restriction, distinct from train's status. (Thread 743236.)

External data, models, or tools carrying licences that restrict commercial use or redistribution can get a winning solution **disqualified and removed from the leaderboard entirely**, not just denied a prize. (Threads 741359, 745072, 745841.)

This matters directly for two notebooks in this map. The MassBank harmonized spectral library used by flexonafft's 2026-09-26 notebook carries its own CC BY-NC-SA 4.0 warning, and the "spectral-library-massbankharmonized" dataset used by nursrijan's apex-pantheon-spectral-titan, jyxxxx's v240-gold-candidate, and lszlst's b432-massbank is filtered to CC BY/CC0 records by two of those three notebooks — but lszlst's b432-massbank ablation deliberately removes that filter to measure the coverage cost, which is exactly the licence exposure the host rulings describe.

Separately, newer public releases of GNPS, MassBank, and MoNA are ruled fine to use, and CC BY-SA records are acceptable (thread 745228). The older, long-standing GNPS-NIST14-MATCHES dataset is grandfathered in, but the host warned that newly created NIST-matched derivative datasets would likely be ruled violations (thread 745832). ICEBERG and GLACIER checkpoints trained on MassSpecGym are fine; NIST-gated checkpoints of the same models are not (threads 742991, 744338). DreaMS weights are allowed (thread 745604).


## What this map does not establish

Several board readings in this map (0.373, 0.384, 0.399, 0.401, 0.390, 0.428) are leaderboard positions observed for a named account on a given date, and the sources are explicit that which notebook version produced each one is not known in most cases. These are included for the timeline, not as notebook-level evidence, and are marked as such above. Several of the highest claimed numbers in the 0.42–0.47 range (rajrajak99's compilations, the MassBank promotion channel's own contribution, synthreaper's v23) have no measured score for the specific step being credited, and are reported here as claims only.
